## 4. Visualizing Churn Drivers & LTV Dynamics
We now visualize the relationship between customer contract commitment, support tickets, and monetary Lifetime Value.

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Plot 1: Churn by Contract Type & Premium Support
sns.barplot(
    data=df, 
    x="contract_type", 
    y="churn", 
    hue="has_premium_support", 
    ax=axes[0], 
    ci=None
)
axes[0].set_title("Churn Rate by Contract Type & Support Tier", fontsize=13)
axes[0].set_ylabel("Mean Churn Rate")
axes[0].set_xlabel("Contract Type")

# Plot 2: LTV Distribution by Churn Status
sns.boxplot(
    data=df, 
    x="churn", 
    y="customer_ltv", 
    ax=axes[1], 
    palette="Set2"
)
axes[1].set_title("Customer Lifetime Value (LTV) Distribution", fontsize=13)
axes[1].set_xticklabels(["Active / Retained", "Churned"])
axes[1].set_ylabel("Customer LTV ($)")

plt.tight_layout()
plt.show()
```
